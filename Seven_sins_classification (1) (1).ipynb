# Chapter 1 — Classifying a player choice
this introduces 20 examples. Each row describes one choice and the consequences.(this is only a dummy of the actual dynamic database would be )
where itll check each weighted consequence and classify it to the deadly sins.
the reason is to go a psychological route on how people would do if a need of good was prompted in front them.(moral compass)

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

## 1. Enter 20 labeled choices
Give each consequence a weight from **0** (absent) to **5** (strong). For example, charging an excessive rescue fee has a high exploitation weight. Destroying a rival’s map can involve both sabotage and retaliation. Lust represents possessive coercion in this game. Neutral covers choices such as keeping the water needed to survive.

```python
features = ["exploitation", "retaliation", "credit_taken", "sabotage",
            "excess_consumption", "duty_abandoned", "possessive_coercion"]

# Each row follows the feature order above, with the choice first and label last.
samples = [
    ["Charge a stranded traveler an excessive rescue fee", 5, 0, 0, 0, 0, 0, 0, "Greed"],
    ["Rig Salman’s scale to overcharge a customer",        4, 0, 0, 0, 0, 0, 0, "Greed"],
    ["Sell donated shelter spaces for personal gain",     5, 0, 0, 0, 0, 1, 0, "Greed"],
    ["Strike a guard after accepting his apology",        0, 5, 0, 0, 0, 0, 0, "Wrath"],
    ["Smash a lantern to avenge an old insult",           0, 4, 0, 1, 0, 0, 0, "Wrath"],
    ["Punish a bandit after Majid accepts surrender",     0, 4, 0, 0, 0, 0, 0, "Wrath"],
    ["Sign Lama’s map as your own discovery",             0, 0, 5, 0, 0, 0, 0, "Pride"],
    ["Take credit for another player’s rescue",           0, 0, 4, 1, 0, 0, 0, "Pride"],
    ["Erase a rival scout’s trail to cost them an award", 0, 0, 0, 5, 0, 0, 0, "Envy"],
    ["Spoil a rival’s harvest after they win praise",     0, 1, 0, 4, 0, 0, 0, "Envy"],
    ["Eat a family’s reserved dates after a full meal",   0, 0, 0, 0, 5, 0, 0, "Gluttony"],
    ["Finish the patrol’s supper despite being full",     0, 0, 0, 0, 4, 1, 0, "Gluttony"],
    ["Skip the watch you accepted while able to serve",   0, 0, 0, 0, 0, 4, 0, "Sloth"],
    ["Leave a promised delivery for others to finish",    0, 0, 0, 0, 0, 5, 0, "Sloth"],
    ["Threaten a host who refuses to give up a keepsake", 0, 1, 0, 0, 0, 0, 5, "Lust"],
    ["Pressure a scout to surrender a personal charm",    0, 0, 0, 0, 0, 0, 4, "Lust"],
    ["Offer a spare water flask to a traveler",           0, 0, 0, 0, 0, 0, 0, "Neutral"],
    ["Keep enough water for your own crossing",           0, 0, 0, 0, 0, 0, 0, "Neutral"],
    ["Credit Lama for finding the shaded route",          0, 0, 0, 0, 0, 0, 0, "Neutral"],
    ["Rest after handing your watch to another guard",    0, 0, 0, 0, 0, 0, 0, "Neutral"]
]

df = pd.DataFrame(samples, columns=["choice"] + features + ["label"])
print("Training samples:", len(df))
df
```

## 2. Train the decision tree
The tree learns from the seven numeric weights (`X`) and the assigned labels (`y`). The choice is an explainations of what they did . All 20 examples are used for this classroom demonstration.

```python
X = df[features]
y = df["label"]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
```

## 3. Trying a custom entry to see if dynamical


```python
custom_choice = "Demand extra coins before letting a caravan use the shade"
custom_weights = {
    "exploitation": 4,
    "retaliation": 0,
    "credit_taken": 0,
    "sabotage": 0,
    "excess_consumption": 0,
    "duty_abandoned": 1,
    "possessive_coercion": 0
}

entry = pd.DataFrame([custom_weights])[features]
prediction = model.predict(entry)[0]
print("Chapter 1 choice:", custom_choice)
print("Predicted label:", prediction)
```
